# Nanopore3 quick start

This notebook is a client of the installed package. It contains no pipeline implementation or hidden state, and works locally or in Colab after the repository is cloned or uploaded.

```python
# Run from the Nanopore3 repository root. In Colab, change this path to the clone.
%pip install -e ".[notebook]"
```

```python
from pathlib import Path
from nanopore3.config import load_config
from nanopore3.pipeline import validate_inputs, run_pipeline
from nanopore3.runtime import doctor_report

doctor_report()
```

```python
config_path = Path("configs/example.yaml")  # replace with your versioned run YAML
config = load_config(config_path)
validate_inputs(config)
```

```python
# This always creates a new immutable run unless an exact run ID is explicitly resumed.
run_directory = run_pipeline(config)
run_directory
```

Open `run_directory / 'stages' / '06_report' / 'report.html'`. For Google Drive, set `output_root` in the YAML to a mounted Drive path. The portable backend requires no MAFFT, SPOA, minimap2, shell setup, or notebook multiprocessing.
